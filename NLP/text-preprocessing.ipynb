# Remove HTML Tag

```python
import re 
def remove_html_tags(text):
    pattern=re.compile('<.*?>')
    return pattern.sub(r'',text)
```

```python
text="""<html>
<head>
  <title>AI and the Future</title>
</head>
<body>
  <h1>Artificial Intelligence</h1>
  <p>Artificial Intelligence (<b>AI</b>) is transforming the way we live and work. 
     It powers everything from <a href="#">voice assistants</a> to self-driving cars.</p>
  <div>
    <h2>Impact on Society</h2>
    <p>AI is expected to <i>revolutionize</i> industries, improve healthcare, and 
       make our lives more efficient.</p>
  </div>
</body>
</html>
"""
```

# Remove URL 

```python
def remove_url(text):
    pattern=re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

# Removing Punctuation

```python
import string,time
string.punctuation
```

```python
exclude=string.punctuation
```

```python
def remove_punc(text):
    for char in exclude:
        text=text.replace(char,'')
    return text
```

```python
text="string. With. Punctuation"
```

```python
start=time.time()
print(remove_punc(text))
time1=time.time()-start
print(time1)
```

```python
def remove_punc1(text):
    return text.translate(str.maketrans(" "," ",exclude))
```

```python
start=time.time()
print(remove_punc1(text))
time2=time.time()-start
print(time2)
```

```python
time1/time2
```

```python
abbr_dict = {}
for line in chat.strip().split("\n"):
    key, value = line.split("=", 1)
    abbr_dict[key.strip()] = value.strip()

print(abbr_dict)
```

```python
def chat_conversation(text):
    new_text = []
    for w in text.split():
        if w.upper() in abbr_dict:
            new_text.append(abbr_dict[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

# Removing Stop Words


```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text=[]
    
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x=new_text[:]
    new_text.clear()
    return " ".join(x)
    
    
```

```python
remove_stopwords("My name is CJ and i Love gaming ")
```

# Stemming


```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample="walk walks walking walked"
stem_words(sample)
```

# Lemmitization

```python
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer=WordNetLemmatizer()

sentence="He was running and eating at same time. He has bat habit of swimming after playing long hours in the sun."
punctuations="?:!.,;"
sentence_words=nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)
sentence_words
print('{0:20}{1:20}'.format("Word","Lemma"))
for word in sentence_words:
    print("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word)))
```
